# 01 — Explore a LeRobot dataset (SIM_PHASE_GUIDE topic 5)

Run with the `chess-arm` kernel. Goal: see the numbers behind what Rerun showed you.

```python
from lerobot.datasets import LeRobotDataset

ds = LeRobotDataset("lerobot/pusht")
print(ds)
```

## Size and timing

```python
print("fps:", ds.fps)
print("episodes:", ds.num_episodes)
print("frames:", ds.num_frames)
```

## Features: what is stored per timestep

```python
for name, spec in ds.features.items():
    print(f"{name:28s} {spec['dtype']:8s} {spec['shape']}")
```

## One frame
`observation.*` is what the policy sees, `action` is what the human commanded at that moment.

```python
frame = ds[0]
for k, v in frame.items():
    print(f"{k:28s}", getattr(v, 'shape', v))
```

## Action vs state over the first steps
The action (target) usually leads the state (where the agent actually is).

```python
for i in range(0, 20, 2):
    f = ds[i]
    print(i, "state", f["observation.state"].tolist(), "action", f["action"].tolist())
```

## Episode boundaries
Episodes are slices of one long table of frames.

```python
ep = ds.meta.episodes[0]
print("episode 0 frames:", ep["dataset_from_index"], "->", ep["dataset_to_index"])
```

## Normalisation statistics
Computed at recording time, used by the processors to scale inputs and outputs.

```python
for k, s in ds.meta.stats.items():
    print(k, {name: getattr(val, 'shape', val) for name, val in s.items()})
```
